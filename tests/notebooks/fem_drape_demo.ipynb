```python
import os
import sys
from pathlib import Path

REPO_ROOT = Path.cwd()
for p in (REPO_ROOT, REPO_ROOT / "tests"):
    if str(p) not in sys.path:
        sys.path.insert(0, str(p))

import numpy as np
from conftest import _resolve_qydp

qydp = _resolve_qydp()
simulator = qydp.simulator
```

```python
from frontend.scenes import load_scene
from frontend.drape_window import make_renderer, run_interactive


def run_secne(scene_name, params=None):
    scene = load_scene(scene_name)
    renderer = make_renderer(scene)
    try:
        run_interactive(renderer, simulator, scene, params)
    finally:
        try:
            renderer.close()
        except Exception as exc:  # noqa: BLE001 - Warp teardown can be brittle
            print(f"renderer close failed: {exc}")

```

```python
scene_name = 'gcd:rand_00YONAPXZE'
params = {
    "smooth_times": 5,
    "step_h": 0.003,
    "sewing_k": 200000.0,
    "query_radius": 0.001,
    "sewing_forced_connect_frame": 80,
    "vf_force_k": 1.5,
    "vf_ground_k": 20,
    "ee_force_k": 0.2,
    "ef_force_k": 0.5,
    "ground": 1,
    "vf_force_type": 0,
    "ee_force_type": 0,
    "max_vel": 100.0,
    "gravity": -9.8,
    "linear_solver_type": 0,
    "pd_iters": 5,
    "pc_iters": 0,
    "subspace_iters": 0,
    "linear_iters": 5,
    "mask_stiff": 2000,
    "max_force_scale": 0.1,
    "average_mass_by_cloth": 0,
    "debug_e_id": -7,
    "debug_v_id": -6,
    "constitutive_model_planar": 1,
    "bending_model": 2,
    "bending_k": 0.01
}
run_secne(scene_name, params)
```
